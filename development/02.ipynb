This notebook provides a simple code to get the `check.py` file from the main branch of the GitHub repository and use it for developing the exercise. Once the `check.py` file is saved to the working directory of a page in a book where the exercise is to be used, you can put it into a directory `exercise_checker` along with an empty `__init__.py` file to use the tool as illustrated in the other example notebooks.

```python
# The code in this cell was written with the help of GitHub Copilot.
import importlib.util
from urllib.parse import quote
url = "https://raw.githubusercontent.com/teachbooks/exercise-checker/main/exercise_checker/check.py"
local_path = "check.py"
quote(url, local_path)

spec = importlib.util.spec_from_file_location("check", local_path)
check = importlib.util.module_from_spec(spec)
spec.loader.exec_module(check)
```
